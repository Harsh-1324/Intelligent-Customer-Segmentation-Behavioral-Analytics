# Intelligent Customer Segmentation & Behavioral Analytics
## Notebook 01: Exploratory Data Analysis (EDA)

### Project Overview
This notebook explores customer behavioral data from **Pheraas**, an event and wedding celebration marketplace in Maharashtra (featuring venues and vendors across Mumbai, Pune, Lonavala, Alibaug, and Nashik).

### Objectives:
1. Inspect dataset structure, missing values, and feature types.
2. Analyze demographic and event-type distributions.
3. Investigate spending, booking frequency, recency, and platform engagement.
4. Examine correlations between numerical features to prepare for clustering.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Styling
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 11
```

### 1. Load Dataset

```python
data_path = os.path.join('..', 'data', 'customer_data.csv')
df = pd.read_csv(data_path)
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns\n")
df.head()
```

### 2. Dataset Information & Missing Value Check

```python
print("--- Summary Info ---")
df.info()

print("\n--- Missing Values Check ---")
print(df.isnull().sum())
```

### 3. Summary Statistics of Key Numerical Features

```python
numerical_cols = [
    'Age', 'Tenure_Months', 'Total_Bookings', 'Total_Spend_INR', 
    'Avg_Booking_Value_INR', 'Days_Since_Last_Inquiry', 'Discount_Usage_Ratio',
    'Website_Visits', 'App_Sessions', 'WhatsApp_Inquiries', 'Cart_Shortlist_Additions'
]
df[numerical_cols].describe().T
```

### 4. Categorical Distributions: Cities and Event Types

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# Location breakdown
city_counts = df['Location'].value_counts()
axes[0].bar(city_counts.index, city_counts.values, color='#2A9D8F')
axes[0].set_title('Customer Distribution by Celebration Destination')
axes[0].set_ylabel('Customer Count')
axes[0].tick_params(axis='x', rotation=45)

# Event Type breakdown
event_counts = df['Primary_Event_Type'].value_counts()
axes[1].bar(event_counts.index, event_counts.values, color='#E76F51')
axes[1].set_title('Distribution of Event Categories')
axes[1].set_ylabel('Customer Count')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

### 5. Spending & Booking Value Distributions

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

sns.histplot(df['Total_Spend_INR'], kde=True, bins=40, color='#1D3557', ax=axes[0])
axes[0].set_title('Total Customer Spend Distribution (₹)')
axes[0].set_xlabel('Total Spend (INR)')

sns.boxplot(x='Primary_Event_Type', y='Total_Spend_INR', data=df, ax=axes[1], palette='Set2')
axes[1].set_title('Total Spend Across Event Categories')
axes[1].set_xlabel('')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

### 6. Behavioral Engagement Signals

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.histplot(df['Days_Since_Last_Inquiry'], kde=True, color='#E63946', ax=axes[0, 0])
axes[0, 0].set_title('Recency: Days Since Last Interaction')

sns.scatterplot(x='Website_Visits', y='Total_Bookings', data=df, alpha=0.5, color='#457B9D', ax=axes[0, 1])
axes[0, 1].set_title('Website Visits vs Total Bookings')

sns.histplot(df['Discount_Usage_Ratio'], kde=True, color='#F4A261', ax=axes[1, 0])
axes[1, 0].set_title('Discount Coupon Usage Ratio')

sns.scatterplot(x='WhatsApp_Inquiries', y='Total_Spend_INR', data=df, alpha=0.5, color='#2A9D8F', ax=axes[1, 1])
axes[1, 1].set_title('WhatsApp Vendor Inquiries vs Total Spend')

plt.tight_layout()
plt.show()
```

### 7. Correlation Analysis

```python
plt.figure(figsize=(12, 8))
corr = df[numerical_cols].corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap='coolwarm', cbar=True, square=True)
plt.title('Correlation Matrix of Behavioral Features')
plt.show()
```

### Key Insights from EDA:
1. **Bimodal Spend Distribution**: Destination Weddings and Corporate events exhibit high ticket sizes (> ₹5,00,000), while Birthdays and local parties cluster under ₹1,00,000.
2. **Discount Dependency**: Price-sensitive customers show high coupon usage (> 60%) coupled with lower booking frequency.
3. **Engagement-Conversion Link**: High WhatsApp inquiries and App sessions strongly correlate with booking conversion.
